```python
df1['EstimatedSalary'].unique()[:10]
```

```python
df1["EstimatedSalary"] = (
    df1["EstimatedSalary"]
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace("- ", "-", regex=False)
    .str.strip()
)

df1["EstimatedSalary"] = df1["EstimatedSalary"].astype(float)
```

```python
df1['EstimatedSalary'].unique()[:10]
```

```python
df2['Balance'].unique()[:10]
```

```python
df2["Balance"] = (
    df2["Balance"]
    .str.replace("€", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace("- ", "-", regex=False)
    .str.strip()
)

df2["Balance"] = pd.to_numeric(df2["Balance"], errors="coerce")
```

```python
df1=df1.drop("Surname",axis=1)
```

```python
df1["Age"] = df1["Age"].fillna(df1["Age"].median())
```

```python
df1.isnull().sum()
```

```python
df1.columns = df1.columns.str.replace('([a-z])([A-Z])', r'\1_\2', regex=True).str.lower()
```

```python
df1.columns
```

```python
df2.columns = df2.columns.str.replace('([a-z])([A-Z])', r'\1_\2', regex=True).str.lower()
```

```python
df2.columns
```

```python
df2['churn_status'] = df2['exited'].map({1: 'Churned', 0: 'Retained'})
```

```python
df2.head(5)
```

```python
df1['geography'] = df1['geography'].replace('FRA', 'France')
```

```python
df = df1.merge(df2, on='customer_id', how='left')
```

```python
df.head(10)
```

```python
df = df.drop('tenure_y',axis=1)
```

```python
df = df.rename(columns={'tenure_x': 'tenure'})
```

```python
df.head(5)
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[18, 30, 45, 60, 100],
    labels=["18-30", "31-45", "46-60", "60+"],
    include_lowest=True
)
```

```python
df["credit_score_group"] = pd.cut(
    df["credit_score"],
    bins=[0, 600, 800, float("inf")],
    labels=["0–599", "600–799", "800+"],
    right=False
)
```

```python
df.head(5)
```

```python
df['geography'] = df['geography'].replace('FRA', 'France')
```

```python
df = df.rename(columns={'geography': 'country'})
```
